**Pertanyaan praktikum no 2**

tampilan tidak berubah tetapi ukuran alokasi ram berbeda kalau int16 itu 16bit / 2byte jika int32 32 bit / 4byte

```python
import numpy as np
import matplotlib.pyplot as plt

canvas_int16 = np.zeros((530, 530, 3), dtype=np.int16)
canvas_int32 = np.zeros((530, 530, 3), dtype=np.int32)

print("=== PERBANDINGAN KANVAS int16 vs int32 ===")
print(f"Canvas int16 -> dtype: {canvas_int16.dtype}, min: {canvas_int16.min()}, max: {canvas_int16.max()}, nbytes: {canvas_int16.nbytes} bytes")
print(f"Canvas int32 -> dtype: {canvas_int32.dtype}, min: {canvas_int32.min()}, max: {canvas_int32.max()}, nbytes: {canvas_int32.nbytes} bytes")

# Menampilkan kedua kanvas
fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(canvas_int16)
axes[0].set_title("int16")
axes[1].imshow(canvas_int32)
axes[1].set_title("int32")
plt.show()
```

**Pertanyaan praktikum no 4**

Keduanya sama" menapikan warna RGB tetapi jika skimage.io.imread() membaca file dalam format RGB, sedangkan ccv2.cvtColor(img, cv2.COLOR_BGR2RGB) mengubah citra BGR bawaan OpenCV menjadi RGB.

```python
import cv2
import skimage.io as skio
import matplotlib.pyplot as plt

# Membaca gambar langsung RGB
img_skimage = skio.imread('KTMCahayaMatahari.jpg')

# Membaca gambar dari cv dan mengkonversinya dari BGR ke RGB
img_cv2 = cv2.imread('KTMCahayaMatahari.jpg')
img_cv2_rgb = cv2.cvtColor(img_cv2, cv2.COLOR_BGR2RGB)

print("=== Pembuktian Nilai Piksel pada Koordinat (100,100) ===")
print("Nilai [R, G, B] skimage :", img_skimage[100, 100])
print("Nilai [R, G, B] OpenCV  :", img_cv2_rgb[100, 100])

fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(img_skimage)
axes[0].set_title("skimage.io.imread()")
axes[1].imshow(img_cv2_rgb)
axes[1].set_title("cv2.cvtColor (BGR2RGB)")
plt.show()
```

**Pertanyaan praktikum no 5**

img.shape tidak ikut berubah karena figsize hanya memperbesar atau memperkecil bingkai tampilan visual di layar tanpa mengubah matriks data piksel di dalamnya, sedangkan perbedaannya cv2.resize secara fisik menambah/mengurangi jumlah piksel pada gambar sehingga merubah dimensi array (img.shape) dan ukuran memorinya sedangkan figsize hanya mengatur skala tampilan gambar pada kanvas Matplotlib.

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('KTMCahayaMatahari.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Kanal Merah-Biru (Kanal Hijau / Index 1 dinolkan)
img_rb = img_rgb.copy()
img_rb[:, :, 1] = 0

# Kanal Hijau-Biru (Kanal Merah / Index 0 dinolkan)
img_gb = img_rgb.copy()
img_gb[:, :, 0] = 0

# Tampilkan berdampingan dalam satu figure
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
fig.suptitle("Kombinasi Kanal Warna", fontsize=14)

axes[0].imshow(img_rb)
axes[0].set_title("Kanal Merah - Biru")
axes[0].axis('off')

axes[1].imshow(img_gb)
axes[1].set_title("Kanal Hijau - Biru")
axes[1].axis('off')

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('KTMCahayaMatahari.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# pencerminan
flip_v = cv2.flip(img_rgb, 0)   # Vertikal
flip_h = cv2.flip(img_rgb, 1)   # Horizontal
flip_both = cv2.flip(img_rgb, -1) # Keduanya (Vertikal & Horizontal)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(flip_v)
axes[0].set_title("Pencerminan Vertikal")
axes[0].axis('off')

axes[1].imshow(flip_h)
axes[1].set_title("Pencerminan Horizontal")
axes[1].axis('off')

axes[2].imshow(flip_both)
axes[2].set_title("Pencerminan Keduanya")
axes[2].axis('off')

plt.show()
```

**Pertanyaannya**

Berapa persen piksel pada foto KTM yang tergolong gelap (nilai intensitas < 100%) ?

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Baca gambar dan mengubah ke grayscale
img_gray = cv2.imread('KTMCahayaMatahari.jpg', cv2.IMREAD_GRAYSCALE)

total_pixel = img_gray.size
piksel_gelap = np.sum(img_gray < 100)
persentase_gelap = (piksel_gelap / total_pixel) * 100

print(f"Total Piksel   : {total_pixel}")
print(f"Piksel Gelap   : {piksel_gelap}")
print(f"Persentase     : {persentase_gelap:.2f}%")

plt.imshow(img_gray < 100, cmap='gray')
plt.title(f"Area Gelap (<100): {persentase_gelap:.2f}%")
plt.axis('off')
plt.show()
```

**Kesimpulan Pertanyaan**

Berdasarkan hasil perhitungan program, foto KTM ini memiliki 28,96% piksel yang tergolong gelap. Bagian gelap ini umumnya berasal dari area teks, bayangan, atau latar belakang foto.

```python
import cv2 as cv
import matplotlib.pyplot as plt

file_foto = ['Warjo.jpeg', 'Ngopi.jpg.jpeg']
teks = "DOKUMENTASI KEGIATAN - PCVK"

hasil_gambar = []

for filename in file_foto:
    img_bgr = cv.imread(filename)
    img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

    # Mengambil tinggi dan lebar
    h, w, _ = img_rgb.shape

    bar_height = int(h * 0.08)         # 8% dari tinggi foto
    font_scale = w / 700               # Menyesuaikan lebar foto
    thickness = max(1, int(w / 400))   # Ketebalan huruf adaptif
    padding_x = int(w * 0.04)          # Jarak teks dari kiri (4%)
    text_y = int(h - (bar_height * 0.3))

    # Buat Bar Label Hitam Padat di Bawah
    pt1 = (0, h - bar_height)
    pt2 = (w, h)
    cv.rectangle(img_rgb, pt1, pt2, color=(0, 0, 0), thickness=-1)

    # Tempel Teks Identitas
    font = cv.FONT_HERSHEY_SIMPLEX
    cv.putText(img_rgb, teks, (padding_x, text_y), font, font_scale,
               color=(255, 255, 255), thickness=thickness, lineType=cv.LINE_AA)

    hasil_gambar.append((filename, img_rgb))

fig, axes = plt.subplots(1, len(hasil_gambar), figsize=(12, 6))

for i, (fn, img_out) in enumerate(hasil_gambar):
    axes[i].imshow(img_out)
    axes[i].set_title(f"{fn}\n{img_out.shape[1]}x{img_out.shape[0]} px")
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```

**Studi Kasus 2**

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

file_foto = ['Warjo.jpeg', 'Ngopi.jpg.jpeg']
nama_toko = "HAPIP STORE"

hasil_thumbnail = []

for filename in file_foto:
    img_bgr = cv.imread(filename)
    img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

    h, w, _ = img_rgb.shape

    # ukuran kanvas persegi (1:1)
    sisi_kanvas = max(h, w)

    # kanvas kosong warna abu-abu terang (RGB: 240, 240, 240)
    kanvas = np.full((sisi_kanvas, sisi_kanvas, 3), 240, dtype=np.uint8)

    # koordinat offset (jarak) agar gambarnya di tengah
    offset_x = (sisi_kanvas - w) // 2
    offset_y = (sisi_kanvas - h) // 2

    kanvas[offset_y:offset_y + h, offset_x:offset_x + w] = img_rgb

    font = cv.FONT_HERSHEY_SIMPLEX
    font_scale = sisi_kanvas / 800
    thickness = max(1, int(sisi_kanvas / 400))
    padding = int(sisi_kanvas * 0.04)

    (text_w, text_h), baseline = cv.getTextSize(nama_toko, font, font_scale, thickness)

    pos_x = sisi_kanvas - text_w - padding
    pos_y = text_h + padding

    pt1_badge = (pos_x - 10, pos_y - text_h - 10)
    pt2_badge = (pos_x + text_w + 10, pos_y + baseline)
    cv.rectangle(kanvas, pt1_badge, pt2_badge, color=(50, 50, 50), thickness=-1)

    cv.putText(kanvas, nama_toko, (pos_x, pos_y), font, font_scale,
               color=(255, 255, 255), thickness=thickness, lineType=cv.LINE_AA)

    hasil_thumbnail.append((filename, kanvas))

fig, axes = plt.subplots(1, len(hasil_thumbnail), figsize=(12, 6))

for i, (fn, img_out) in enumerate(hasil_thumbnail):
    axes[i].imshow(img_out)
    axes[i].set_title(f"Thumbnail 1:1 - {fn}\nDimensi: {img_out.shape[1]} x {img_out.shape[0]} px")
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```
